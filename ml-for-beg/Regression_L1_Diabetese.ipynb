```python
import matplotlib.pyplot as plt
import numpy as np 
from sklearn import datasets, linear_model, model_selection

X,y = datasets.load_diabetes(return_X_y = True) #Unpacking of data, it returns data, and target

""" same concept but implemnetation is basic just for the understanding 
result = datasets.load_diabetes(return_X_y=True)
X = result[0]
y = result[1]
"""

print(X.shape) # returns just like size of the list or array
print(X[0])
```

```python
X = X[:, 2] #gets the BMI values copy in this column and throw the others away
print(X.shape) 

#Reshape to a 2D array 
X = X.reshape((-1, 1))
print(X.shape)
```

```python
X_train, X_test, y_train, y_test = model_selection.train_test_split(X, y, test_size=0.33)
```

```python
# create a linear regression model
model = linear_model.LinearRegression()
model.fit(X_train, y_train)
```

```python
y_pred = model.predict(X_test)
```

```python
plt.scatter(X_test, y_test, color= 'black')

plt.plot(X_test, y_pred, color= 'blue' , linewidth=3)

plt.xlabel('Scaled BMIs')
plt.ylabel('Disease Progression')
plt.title('A graph plot showing Diabetese Progression Against BMI')

plt.show()
```
